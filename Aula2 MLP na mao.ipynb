```python
# exemplo basico de um unico neuronio
inputs = [0.5, 0.13, 0.9]
pesos = [0.4, 0.12, 0]
bias = 2

output = inputs[0] * pesos[0] + inputs[1] * pesos[1] + inputs[2] * pesos[2] + bias
print(output)

# https://numpy.org/devdocs/reference/generated/numpy.dot.html

output2 = np.dot(pesos, inputs) + bias
print(output2)
```

```python
inputs = [0.5, 0.13, 0.9]

pesos = [[0.6, 0.12, 0.6], [-0.4, 0.18, 0.1], [0.9, -0.12, 0.3]]

bias = [2, 3, -4]

outputs = []
# https://docs.python.org/3.3/library/functions.html#zip
for npesos, nbias in zip(pesos, bias):
    noutput = 0
    for ninput, peso in zip(inputs, npesos):
        noutput = noutput + (ninput * peso)
    noutput = noutput + nbias
    # https://www.w3schools.com/python/ref_list_append.asp
    outputs.append(noutput)

print(outputs)
```
